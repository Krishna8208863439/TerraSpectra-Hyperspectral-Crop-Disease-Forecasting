# Day 1: Environment Verification & Initial Dataset Inspection

This notebook verifies that the core scientific Python stack (`numpy`, `pandas`, `scikit-learn`, `xgboost`, `matplotlib`, `spectral`) imports cleanly and inspects the selected dataset (`data/raw/Multispectral_Reflectance.csv` + `data/raw/experimental_design.csv`) for shape, labels, and grouping columns (`Plant_ID`, `Pot_Number`, `date`).

```python
import numpy as np
import pandas as pd
import sklearn
import xgboost as xgb
import matplotlib.pyplot as plt
import spectral

print('NumPy:', np.__version__)
print('pandas:', pd.__version__)
print('scikit-learn:', sklearn.__version__)
print('XGBoost:', xgb.__version__)
print('matplotlib:', plt.matplotlib.__version__)
print('spectral:', spectral.__version__)
```

```python
design = pd.read_csv('../data/raw/experimental_design.csv')
spectra = pd.read_csv('../data/raw/Multispectral_Reflectance.csv')

df = spectra.merge(design, on='Plant_ID', how='left')
print('Merged shape:', df.shape)
print('Treatments / Labels:\n', df['Treatment'].value_counts(dropna=False))
print('Unique Plants:', df['Plant_ID'].nunique(), '| Unique Dates:', df['date'].nunique())
```

```python
band_cols = ['Mean_365', 'Mean_460', 'Mean_525', 'Mean_570', 'Mean_645', 'Mean_670', 'Mean_700', 'Mean_780', 'Mean_890', 'Mean_970']
wavelengths = [int(c.split('_')[1]) for c in band_cols]

fig, ax = plt.subplots(figsize=(9, 4.5))
for label, grp in df.groupby('Treatment'):
    ax.plot(wavelengths, grp[band_cols].mean(), marker='o', label=label)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Mean Reflectance (%)')
ax.set_title('Mean Leaf/Canopy Spectral Reflectance by Treatment')
ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
